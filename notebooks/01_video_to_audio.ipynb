```python
import os

files = os.listdir("../data/Videos")
for file in files:
    print(file)
    print("====================="*3)
```

```python
import subprocess

for file in files:
    file_num = file.split("#")[1].split(".")[0]
    print("File Number:", file_num)
    file_name = file.split("｜")[0]
    print("File name:", file_name,"\n")
    subprocess.run(["ffmpeg", "-i", f"../data/Videos/{file}", f"../data/audios/{file_num}_{file_name}.mp3"])
```
