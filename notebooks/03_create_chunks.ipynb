```python
import whisper
import json
import os
```

```python
model = whisper.load_model("large-v2")
```

```python
audios = os.listdir("/content/drive/MyDrive/RAG Based AI Teaching Assistant/data/audios")
```

```python
for audio in audios:
  number = audio.split("_")[0]
  title = audio.split("_")[1][:-4]
  # print(number,"\n", title)

  result = model.transcribe(audio=f"/content/drive/MyDrive/RAG Based AI Teaching Assistant/data/audios/{audio}",
                            language="hi",
                            task="translate",
                            word_timestamps=False)
  chunks = []
  for segment in result["segments"]:
    chunks.append({"number": number, "title": title, "start": segment["start"], "end": segment["end"], "text": segment["text"]})

  chunks_with_metadata = {"chunks": chunks, "text": result["text"]}

  with open(f"/content/drive/MyDrive/RAG Based AI Teaching Assistant/data/jsons/{audio}.json", "w") as f:
    json.dump(chunks_with_metadata, f)
```
